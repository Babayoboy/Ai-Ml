```python
# Raw Data

df = pd.read_csv("raw_data.csv")
```

```python
df.drop_duplicates(inplace=True)

df = df.fillna(0)
```

```python
# Feature Engineering
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x: "20%" if x >= 50000 else "10%")
gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)

df2 = df2.assign(new_income = df2["income"] * 1.1)

df2["country"] = df2["country"].replace("USA", "US")

df2.columns = [column.capitalize() for column in df2.columns]
df2 = df2.rename(columns={"Income":"Salary"})
df2 = df2.rename(index={1:"First"})

# df2["Salary"].sort_values()
df2 = df2.sort_values(["Salary", "Age"], ascending=False)
df2 = df2.reset_index(drop=True)
```

```python
# Grouping and Aggregation
df2.groupby("Country")["Salary"].mean()
df2.groupby("Country")["Salary"].sum()
df2.groupby("Gender")["Salary"].max()
df2.groupby("Gender")["Salary"].count()

df2.groupby("Country")["Salary"].agg(["mean", "sum", "max","min", "count"])

df2.groupby("Gender")["Salary"].agg(avg_salary = "mean",sum_salary = "sum",max_salary = "max",min_salary = "min",count = "count")

df2.groupby("Country").agg({
    "Salary" : "mean",
    "Age": "max"
})
df2.groupby("Country").agg(
    avg_salary = ("Salary", "mean"),
    max_age = ("Age", "max")
)
```
