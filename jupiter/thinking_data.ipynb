```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        return json.load(f)
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    text_to_num = {"one": 1, "two":2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = float(raw_rating)

        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
        
```

```python
data = clean_data(data)
print(data)
```

```python
def avg_rating(data):
    sum_rating = 0
    count = 0
    for user in data:
        sum_rating += float(user["rating"])
        count += 1

    return round(sum_rating / count, 1)
        
```

```python
avg_rating(data)
```

```python
def poor_rating(data):
    count_user = 0
    count_poor_rating = 0
    for user in data:
        count_user += 1
        if (user["rating"] < 3):
            count_poor_rating += 1
    return f"{round((count_poor_rating/count_user) * 100, 1)}%"
    
```

```python
poor_rating(data)
```

```python
def recommend(data):
    recommendation = []
    for user in data:
        curr_recomend = {}
        curr_recomend["name"] = user["name"]
        if (user["rating"] >= 4):
            curr_recomend["recomendation"] = "Apple"
        else:
            curr_recomend["recomendation"] = "Samsung"
        recommendation.append(curr_recomend)
    return recommendation
```

```python
recommend(data)
```
